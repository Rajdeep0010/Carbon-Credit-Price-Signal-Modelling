## ***COP PREPARATION***

```python
cop_dates = pd.to_datetime([
    '2015-11-30',   # COP21 Paris
    '2016-11-07',   # COP22 Marrakech
    '2017-11-06',   # COP23 Bonn
    '2018-12-02',   # COP24 Katowice
    '2019-12-02',   # COP25 Madrid
    # COP26 was postponed from 2020 to 2021 due to COVID — no 2020 entry
    '2021-11-01',   # COP26 Glasgow
    '2022-11-07',   # COP27 Sharm el-Sheikh
    '2023-11-30',   # COP28 Dubai
    '2024-11-11',   # COP29 Baku
    '2025-11-10',   # COP30 Belem Brazil
])

print("Total COP summits in range:", len(cop_dates))
print(cop_dates)
```

```python
#weekly date spine
weekly_dates = pd.date_range(start='2015-01-01', end='2026-04-30', freq='W')

cop_df = pd.DataFrame(index=weekly_dates)
cop_df.index.name = 'date'

print("Total weeks:", len(cop_df))
cop_df.head()
```

```python
def days_to_next_cop(date):
    future_cops = cop_dates[cop_dates >= date]
    
    if len(future_cops) == 0:
        # After COP30 — no future COP in range
        # Assign 365 as neutral placeholder
        return 365
    
    return (future_cops[0] - date).days

cop_df['days_to_next_cop'] = [days_to_next_cop(d) for d in cop_df.index]

print("Days to next COP calculated")
cop_df.head(10)
```

```python
print("Shape:", cop_df.shape)
print()
print("Sample around COP30 Belem (Nov 2025):")
print(cop_df['2025-10-01':'2025-12-01'])
```

```python
print("days_to_next_cop stats:")
print(cop_df['days_to_next_cop'].describe())
print()
print("Weeks where COP is within 30 days:", 
    (cop_df['days_to_next_cop'] <= 30).sum())
print("Weeks after COP30 with no future COP (value=365):", 
    (cop_df['days_to_next_cop'] == 365).sum())
```

```python
# Ensure date is the index for every dataframe
# If date is already the index this does nothing harmful

def ensure_date_index(df, date_col='date'):
    if date_col in df.columns:
        df = df.set_index(date_col)
    df.index = pd.to_datetime(df.index)
    df.index.name = 'date'
    return df

carbon = ensure_date_index(carbon)
gas    = ensure_date_index(gas)
co2    = ensure_date_index(co2)
vix    = ensure_date_index(vix)

# Confirm all indexes are clean
print("carbon index:", carbon.index.name, "| dtype:", carbon.index.dtype)
print("gas    index:", gas.index.name,    "| dtype:", gas.index.dtype)
print("co2    index:", co2.index.name,    "| dtype:", co2.index.dtype)
print("vix    index:", vix.index.name,    "| dtype:", vix.index.dtype)
print("cop_df index:", cop_df.index.name, "| dtype:", cop_df.index.dtype)
```

```python
df = cop_df.copy()
df = df.join(carbon,  how='left')
df = df.join(gas,     how='left')
df = df.join(co2,     how='left')
df = df.join(vix,     how='left')

print("After joining all datasets:", df.shape)
print("\nMissing values per column:")
print(df.isnull().sum())
df.head()
```

```python
print("Gas dataframe date samples:")
print(gas.index[:10])
print()
print("Weekly spine date samples:")
print(cop_df.index[:10])
```

```python
# Henry Hub has occasional gaps on holidays
# Forward fill maximum 1 week to avoid fabricating data
df['gas_price_weekly'] = df['gas_price_weekly'].ffill(limit=1)

print("Gas price nulls after ffill:", df['gas_price_weekly'].isnull().sum())
display(df.sample(10))
```
